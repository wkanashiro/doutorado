```python
import os
from pathlib import Path
from tqdm import tqdm 
```

```python
dataset_root = "/media/guatambu/hdd/wesley/soja_patches_reduced_dino/"

print(f"--- Renomeando arquivos em {dataset_root} para formato numérico ---")
    
# Processa TRAIN e VAL separadamente para manter a organização
for split in ['train', 'val']:
    split_path = os.path.join(dataset_root, split)
    if not os.path.exists(split_path):
        continue
        
    # Percorre cada classe (pasta) dentro do split
    for class_dir in os.listdir(split_path):
        full_class_dir = os.path.join(split_path, class_dir)
        if not os.path.isdir(full_class_dir):
            continue
            
        print(f"Processando pasta: {split}/{class_dir}")
        
        # Lista todos os arquivos atuais
        files = [f for f in Path(full_class_dir).glob("*") if f.is_file()]
        
        # Renomeia sequencialmente: 0.JPEG, 1.JPEG, 2.JPEG...
        for i, file_path in enumerate(tqdm(files)):
            # Define novo nome
            new_name = f"{class_dir}_{i}.JPEG"
            new_path = file_path.parent / new_name
            
            # Evita sobrescrever se o arquivo já tiver o nome certo
            if file_path != new_path:
                try:
                    file_path.rename(new_path)
                except OSError as e:
                    print(f"Erro ao renomear {file_path}: {e}")

print("\nConcluído! Todos os arquivos agora são números (ex: 10.JPEG).")
```

## Experimento 2 - 20% do dataset original // 95% train e 5% val

```python
# Criar pasta cópia com 1220000 de imagens do dataset original de forma aleatoria
import random
import shutil
import os
from pathlib import Path
from tqdm import tqdm 

caminho_origem = "/media/guatambu/hdd/wesley/patches/"
caminho_destino = "/media/guatambu/hdd/wesley/soja_patches_6M/"

# Como são todos os arquivos, não preciso selecionar aleatoriamente, posso pegar os primeiros 6101300

if not os.path.exists(caminho_destino):
    os.makedirs(caminho_destino)
todos_arquivos = [f for f in os.listdir(caminho_origem) if os.path.isfile(os.path.join(caminho_origem, f))]
amostra_aleatoria = todos_arquivos[:6101300]

for arquivo in tqdm(amostra_aleatoria):
    shutil.copy(os.path.join(caminho_origem, arquivo), os.path.join(caminho_destino, arquivo))
    
```

```python
# Subdividir a pasta em TRAIN e VAL (95% TRAIN, 5% VAL)
caminho_base = "/media/guatambu/hdd/wesley/soja_patches_6M/"
caminho_train = os.path.join(caminho_base, "train", "soja")
caminho_val = os.path.join(caminho_base, "val", "soja")

if not os.path.exists(caminho_train):
    os.makedirs(caminho_train)
if not os.path.exists(caminho_val):
    os.makedirs(caminho_val)
todos_arquivos = [f for f in os.listdir(caminho_base) if os.path.isfile(os.path.join(caminho_base, f))]
random.shuffle(todos_arquivos)

limite_train = int(0.95 * len(todos_arquivos))
train_files = todos_arquivos[:limite_train]
val_files = todos_arquivos[limite_train:]

# Renomear e mover os arquivos para as pastas TRAIN e VAL
for i, arquivo in enumerate(tqdm(train_files)):
    new_name = f"soja_{i}.JPEG"
    shutil.move(os.path.join(caminho_base, arquivo), os.path.join(caminho_train, new_name))
    
for i, arquivo in enumerate(tqdm(val_files)):
    new_name = f"soja_{i}.JPEG"
    shutil.move(os.path.join(caminho_base, arquivo), os.path.join(caminho_val, new_name))
```
